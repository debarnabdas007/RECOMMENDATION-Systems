```python
## HELPER Function
import ast
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name'])  # i['name'] gives the key->'name' in the dictionary for every iteration
    return L 

## The genre input is a STRING
## ast.literal_eval(text) converts this into a python object
"""[
    {"id": 28, "name": "Action"},
    {"id": 12, "name": "Adventure"}
]""" 
## This is now a list of dictionaries.. so loop go through the list, not the string
"""
First iteration:
i = {"id": 28, "name": "Action"}
Second:
i = {"id": 12, "name": "Adventure"}
"""
```

```python
movies.dropna(inplace=True) ## Defensive step
```

```python
# convert 'genres'
movies['genres'] = movies['genres'].apply(convert)
# Take every value/row in the genres column, pass it through convert(), and replace the column with the results
```

```python
#$ Fetch the Director ONLY !!
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L
```

```python
## Transformation for removing the space
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
movies['cast'] = movies['cast'].apply(collapse)
movies['crew'] = movies['crew'].apply(collapse)
movies['genres'] = movies['genres'].apply(collapse)
movies['keywords'] = movies['keywords'].apply(collapse)
```

```python
# concatenates the lists element-by-element for each movie
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']


""""
one combined representation of everything that describes the movie

Example:

overview  = [space, alien, planet]
genres    = [action, sci-fi]
keywords  = [future, war]
cast      = [SamWorthington]
crew      = [JamesCameron]

becomes:

tags =
[space, alien, planet, action, sci-fi, future, war,
 SamWorthington, JamesCameron]

Then this single tags field becomes the input to the vectorization step.

"""
```

##### LEmmatization

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
## Examole of STemming
print(ps.stem('loving'))
print(ps.stem('love'))
print(ps.stem('dancing'))
print(ps.stem('dance'))
print(ps.stem('danced'))
```

```python
new['tags'] = new['tags'].apply(stem)
```

```python
new
```

## --- VECTORIZATION ---

#### Bag of Words vectorization

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(
    max_features=5000,  # 5,000 most useful/frequent vocabulary terms
    stop_words='english' # Remove common English words like: the, is, a, an, and, of, in...
)  
```

```python
vector = cv.fit_transform(new['tags']).toarray() ## outputting an ARRAY !!

# fit       → learn vocabulary from all tags
# transform → convert each tag into its numerical vector
```

```python
vector
# Each movie becomes a vector representing its word-count composition
```

```python

vector.shape
```

```python
# Tell me what each column in my vector means
cv.get_feature_names_out()
# he actual vocabulary/features that CountVectorizer learned
```

#### Search

```python

from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vector)
```

```python
similarity
## Diagonally there will be all 1
```

```python
similarity.shape

"""calculates:

Movie 1 ↔ Movie 1
Movie 1 ↔ Movie 2
Movie 1 ↔ Movie 3
...
Movie 4806 ↔ Movie 4806

So you get:

             Movies
          4806 columns
       ┌─────────────────
Movie 1│ 1   .4  .2 ...
Movie 2│ .4  1   .7 ...
Movie 3│ .2  .7  1  ...
  ...  │
Movie 4806
       └─────────────────
        4806 rows

Hence:

$$ \boxed{4806 \times 4806} $$
Important distinction
vector
→ (4806, 5000)

Each movie → 5000-dimensional feature vector

while:

similarity
→ (4806, 4806)

Each movie → similarity score with every movie."""
```

```python
similarity[0][1] 
# similarity[i][j] means: How similar is movie i to movie j
```

```python
similarity[2]
# similarity between 2 index movie with all other movie's vectors
## obviously that 1 down there is for the same same movie comparison
```

```python
# Main Function
def recommend(movie):
    index = new[new['title'] == movie].index[0]  # movies index
    distances = sorted(
        list(enumerate(similarity[index])),  
        reverse=True,
        key = lambda x: x[1]
    )
    
    for i in distances[1:6]:
        print(new.iloc[i[0]].title)
```

```python

recommend('Gandhi')
```
